```python
df = pd.read_csv(r"C:\Users\Huzi\Downloads\ride_hailing_transactions.csv")
df.head(3)
```

```python
df['Items'] = df['Items'].apply(lambda x: x.split(','))
```

```python
from collections import Counter
all_items = [item for basket in df['Items'] for item in basket]
freq = Counter(all_items)
print(freq)
```

```python
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_array = te.fit(df['Items']).transform(df['Items'])
df_encoded = pd.DataFrame(te_array, columns=te.columns_)
```

```python
from mlxtend.frequent_patterns import fpgrowth, association_rules

for min_sup in [0.02, 0.05, 0.08]:
    freq_items = fpgrowth(df_encoded, min_support=min_sup, use_colnames=True)
    print(f"min_support={min_sup} → {len(freq_items)} itemsets found")
```

```python
freq_items = fpgrowth(df_encoded, min_support=0.03, use_colnames=True)
rules = association_rules(freq_items, metric="lift", min_threshold=1.2)
rules = rules.sort_values('lift', ascending=False)
```
